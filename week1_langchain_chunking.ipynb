The reason you don't see overlap between every chunk in RecursiveCharacterTextSplitter comes down to one key rule:

RecursiveCharacterTextSplitter refuses to cut words or sentences in half unless it has no other choice.

1. Why the overlap didn't appear between Chunk 3 and Chunk 4:
In your code, you set:

chunk_size = 150
chunk_overlap = 30
separators = ["\n\n", "\n", ". ", " ", ""]
Look at Chunk 3:

". This includes eight planets, dwarf planets, moons, asteroids, comets, and enormous amounts of dust and gas" $\rightarrow$ Length: 108 characters

When LangChain finishes Chunk 3 and prepares Chunk 4:

It looks back at the end of Chunk 3 to see what it can copy over into Chunk 4.
Your allowed overlap window is only 30 characters.
But the atomic piece at the end of Chunk 3 is a sentence that is 108 characters long ($108 > 30$).
Because RecursiveCharacterTextSplitter prioritizes clean sentence and word boundaries, it will not slice the last 30 characters of that sentence in half (which would look like: "... amounts of dust and gas").
Since the entire sentence cannot fit into the 30-character overlap budget, it drops the overlap completely and cleanly starts Chunk 4 at the next sentence.
2. Why you DO see overlap in Cell 16 (SentenceTransformersTokenTextSplitter):
In your token-based chunker:

Chunk 2 ends with:
"...mercury is the closest planet to the sun"

Chunk 3 starts with:
"rocky. mercury is the closest planet to the sun and has almost no atmosphere..."

Token splitters split at the token (word/subword) level, where every word is only 1–2 tokens. A 10-token overlap can easily grab 7–8 full words, so the overlap appears every single time.

3. How to get consistent overlap in RecursiveCharacterTextSplitter:
Solution A: Increase chunk_overlap
Make chunk_overlap large enough to fit an entire sentence

```python
#5 SEMANTIC CHUNKING

from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEmbeddings

```

```python
text="""

Data privacy is an important consideration when applications process personal information. 
Companies should collect only the information they need, protect stored data, and control who can access it. Users should also understand how their information is collected and used."""
```

```python
embeddings=HuggingFaceEmbeddings(model="sentence-transformers/all-MiniLM-L6-v2")
```

```python
chunker=SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="percentile",#thresholld to determine the brekapoints,percentile or absolute
    #breakpoint_threshold=0.5,#value to determinine the breakpoints

)

chunks=chunker.split_text(text)

for i ,chunk in enumerate(chunks):
    print(f"\nChunk {i+1}")
    print(chunk)
```

```python
splitter=RecursiveCharacterTextSplitter(
    chunk_size=250,
    chunk_overlap=50,
    separators=["\n\n","\n",", "," ",""],
    length_function=len #We tell it to count characters, not words
)
```

```python
chunks=splitter.split_documents(pages)
chunks
```

```python
for i ,chunk in enumerate(chunks):
    print(f"\nchunk {i+1}")
    print(chunk)
```

```python
for i,chunk in enumerate(chunks):
    chunk.metadata["chunk_id"]=f"CHK--{i+1}"
    chunk.metadata["project_name"]="new project"

    if "BUDGET" in chunk.page_content:
        chunk.metadata["category"]="Financials"
    elif "COMPLIANCE" in chunk.page_content:
        chunk.metadata["category"]="Complaint"
    elif "INGEstion" in chunk.page_content:
        chunk.metadata["category"]="Ingestion"
    else:
        chunk.metadata["category"]="general"
```

```python
for chunk in chunks[:4]:
    print(f"\n[{chunk.metadata['chunk_id']}] Category: {chunk.metadata['category']}")
    print("-" * 40)
    print(f"Text Content:\n{chunk.page_content}")
    print("-" * 40)
    print(f"Full Attached Metadata: {chunk.metadata}")
```
